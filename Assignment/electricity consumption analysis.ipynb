```python
#importing numpy
import numpy as np
```

```python
#1. dataset
np.random.seed(42)
consumption = np.random.randint(200,600, size=120)
print("consumption: ", consumption)
```

```python
#2. Basic Statistics
mean_consumption = np.mean(consumption)
median_consumption = np.median(consumption)
min_consumption = np.min(consumption)
max_consumption = np.max(consumption)
std_consumption = np.std(consumption)

print("Mean Consumption:", (round (mean_consumption, 2)))
print("Median Consumption:", median_consumption)
print("Minimum Consumption:", min_consumption)
print("Maximum Consumption:", max_consumption)
print("Standard Deviation:", (round (std_consumption, 2)))
```

```python
#3. Identify Extremes
highest_day_of_consumption = np.where(consumption == np.max(consumption))[0] + 1
lowest_day_of_consumption = np.where(consumption == np.min(consumption))[0] + 1

print("Highest Day Of Consumption:", highest_day_of_consumption)
print("Lowest Day Of Consumption:", lowest_day_of_consumption)
```

```python
#4. Monthly Splitting
months = np.array_split(consumption, 4)
per_month = [print(months) for months in months]
```

```python
#5. Monthly Analysis
print("Total Consumption")
total_consumption = [print(month.sum()) for month in months]
print("\nAverage Daily Consumption")
avg_daily_consumption = [print(round(month.mean(), 2)) for month in months]
```

```python
#6. Weekly Analysis
weeks = np.array_split(consumption, 7)
per_weeks = [print(week) for week in weeks]
```

```python
#7. Overall Consumption Trend
print("Weekly Totals")
weekly_totals = [print(week.sum()) for week in weeks]
print("\nWeekly Averages")
weekly_avgs = [print(round(week.mean(), 2)) for week in weeks]
```

```python
#8. Anomaly Detection
cumulative_consumption = np.cumsum(consumption)

print("First 10 Cummulative Consumption:")
print(cumulative_consumption[:10])
```

```python
lower_bound = mean_consumption - 2 * std_consumption
upper_bound = mean_consumption + 2 * std_consumption
print(round(lower_bound,2))
print(round(upper_bound,2))
anomaly = np.where((consumption < lower_bound) | (consumption > upper_bound))[0] + 1
print("Anomaly Detection (low/high:)", anomaly)
```

Summary
The data is contributed with daily values between 200kwh and 600kwh.
The basic statistic such as mean, median, maximum, minimum and standard deviation are calculated to understand the data.
The consumption data is divided into months and weeks to known the total and average of how the electricity was used.
The cumulative electricity was used to calculatethe consumption during the 120 days. Anomaly detection was used to identify the days when the electricity consumption is high or low which is compared with the use of average.
