```python
import numpy as np
```

```python
#1. Generate statistics
np.random.seed(42) #reproducibility
consumption = np.random.randint(200, 600, size=120)
print("Consumption:", consumption)
```

```python
#2. Basic statistics
total_consumption = np.sum(consumption)
mean_consumption = np.mean(consumption)
median_consumption = np.median(consumption)
min_consumption = np.min(consumption)
max_consumption = np.max(consumption)
std_consumption = np.std(consumption) 

print("Total consumption:", total_consumption)
print("Average consumption:", (round(mean_consumption, 2)))
print("Median consumption:", median_consumption)
print("Minimun consumption:", min_consumption)
print("Maximum consumption:", max_consumption)
print("Standard deviation:", (round(std_consumption, 2)))
```

```python
#3. Identify extremes
day_highest_consumption = np.where(consumption == np.max(consumption))[0] + 1
day_lowest_consumption = np.where(consumption == np.min(consumption))[0] + 1

print("Day of highest consumtion:", day_highest_consumption)
print("Day of lowest consumtion:", day_lowest_consumption)
```

```python
#4. Monthly spliting
months = np.array_split(consumption, 4)
each_month = [print(month) for month in months]
```

```python
#5. Monthly Analysis
print("Monthly Total")
monthly_total = [print(month.sum()) for month in months]

print("\nMonthly Average")
monthly_avg = [print(round(month.mean(), 2)) for month in months]
```

```python
#6. Weekly Analysis
each_week = [month[:28].reshape(4, 7) for month in months]

print("Weekly Total")
weekly_total = [print(week.sum(axis=1)) for week in each_week]

print("\nWeekly Average")
weekly_avg = [print(week.mean(axis=1)) for week in each_week]
```

```python
#7. Overall Consumption
cumulative_consumption = np.cumsum(consumption)

print("First 10 cumulative consumption values:")
print(cumulative_consumption[:10])
```

```python
#8. Anomaly Detection
lower_bound = mean_consumption - 2 * std_consumption
upper_bound = mean_consumption + 2 * std_consumption

print(round(lower_bound, 2))
print(round(upper_bound, 2))

anomaly = np.where((consumption < lower_bound) | (consumption > upper_bound)) [0] + 1
print("Anomaly detection (low/high):", anomaly)
```
