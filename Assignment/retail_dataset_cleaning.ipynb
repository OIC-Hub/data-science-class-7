**Cleaning Dataset**

```python
pd.set_option('display.max_rows', None)
df
```

```python
df[df["Product_ID"] == "P005"]
```

```python
df.loc[18, "Product_ID"] = "P019"
```

```python
df["Product_Name"] = (
    df["Product_Name"]
    .str.strip()
    .str.title()
    .tolist()
)
df["Product_Name"]
```

```python
df["Category"] = (
    df["Category"]
    .str.strip()
    .str.title()
    .tolist()
)
df["Category"]
```

```python
df["Supplier"] = (
    df["Supplier"]
    .str.strip()
    .str.title()
    .tolist()
)
df["Supplier"]
```

```python
df
```

```python
df["Quantity"].to_list()
```

```python
def clean_quantity(value):
    if value == "twenty":
        return 20
    elif value == "ten":
        return 10
    elif isinstance(value, (int, float)):
        return value
    elif isinstance(value, str) and value.isdigit():
        return int(value)
    else: 
        return np.nan

df["Quantity"] = df["Quantity"].apply(clean_quantity)
df["Quantity"] = df["Quantity"].abs()
df["Quantity"]
```

```python
df["Unit_Price"].to_list()
```

```python
df["Unit_Price"] = (
    df["Unit_Price"]
    .astype(str)
    .replace("Nan", np.nan)
    .astype(float)
)
df["Unit_Price"]
```

```python
#removing rows with unit price less than 0
df = df[df["Unit_Price"] >= 0]
```

```python
df
```

```python
from dateutil import parser

def clean_date(date_str):
    try:
        return parser.parse(date_str, dayfirst=True)
    except Exception:
        return pd.NaT

df["Last_Update"] = df["Last_Update"].apply(clean_date)
df["Last_Update"]
```

```python
df
```

```python
df.info()
```

```python
df["Product_Name"] = df["Product_Name"].fillna("Unknown")

df["Quantity"] = df["Quantity"].fillna(df["Quantity"].mode()[0])

df["Last_Update"] = df["Last_Update"].fillna(df["Last_Update"].mode()[0])
```

```python
df
```

```python
df.describe()
```

```python
df["Supplier"] = df["Supplier"].replace("", "Unknown")
```

**PART 4: ANALYSIS AND SORTING**

```python
#1 Which products need restocking (where Quantity < Reorder_Level)?
df[df["Quantity"] < df["Reorder_Level"]]
```

```python
#What is the average unit price per category?
df["Unit_Price"] = pd.to_numeric(df["Unit_Price"], errors="coerce")

df.groupby("Category")["Unit_Price"].mean()
```

```python
#Which supplier provides the most products?
df["Supplier"].value_counts()
```

```python
#List the top 5 most expensive products.
df.sort_values(["Unit_Price"], ascending=False).head()
```

```python
#Sort the cleaned dataset by Category and then by Unit_Price in descending order.
df.sort_values(["Category", "Unit_Price"], ascending=[True,False])
```
