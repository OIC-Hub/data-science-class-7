```python
#cleaning the data
df.dropna()
```

```python
df.rename(columns={
    "Product_ID": "Product Id",
    "Product_Name": "Product Name",
    "Unit_Price": "Unit Price",
    "Recorder_Level": "Recorder Level",
    "Last_Update": "Last Update"
})
```

```python
df["Product_Name"] = (
    df["Product_Name"]
    .str.strip()
    .str.title()
    .str.replace(r"[\s-]+", " ", regex=True)
    .tolist()
)

df["Product_Name"].to_list()
```

```python
df["Category"] = (
     df["Category"]
    .str.strip()
    .str.title()
    .str.replace(r"[\s-]+", " ", regex=True)
    .tolist()
)

df["Category"].to_list()
```

```python
df.replace({
    "Compu Store": "Computer Store",
    "Techworld": "Tech World",
    "CompuStore": "Computer Store",
    "Officehub": "Office Hub",
    "Compustore": "Computer Store"
})
```

```python
def clean_quantity(value):
    if value == "thirty":
        return 30
    elif isinstance(value, (int, float)):
        return value
    elif isinstance(value, str) and value.isdigit():
        return int(value)
    else:
        return np.nan
df["Quantity"] = df["Quantity"].apply(clean_quantity)

df["Quantity"]
```

```python
df["Unit_Price"] = (
    df["Unit_Price"]
    .astype(str)
    .replace('[-,]', "", regex=True)
    .replace("NaN", np.nan)
    .astype(float)
)

df["Unit_Price"]
```

```python
df.drop_duplicates()
```

```python
from dateutil import parser

def clean_date(date_str):
    try:
        return parser.parse(date_str, dayfirst=True)
    except Exception:
        return pd.NaT

df["Last_Update"] = df["Last_Update"].apply(clean_date)
df["Last_Update"]
```

```python
#Analysising the data
restocking = df[df["Quantity"] < df["Reorder_Level"]]

restocking
```
