```python
#import libraries
import numpy as np
import pandas as pd

pd.set_option("display.max_rows", None)
```

```python
np.random.seed(42) #reproducibility
days = np.arange(1, 121)
consumption = np.random.randint(200, 601, size=120)

print("Days:", days)
print("consumption:", consumption)
```

```python
data = pd.DataFrame({
    "Days": days,
    "consumption": consumption
})

data
```

```python
mean_consumption = np.mean(consumption)
std_consumption = np.std(consumption)
print("Mean consumption:", mean_consumption)
print("Median consumption:", np.median(consumption))
print("Minimum consumption:", np.min(consumption))
print("Maximum consumption:", np.max(consumption))
print("Standard deviation consumption:", std_consumption)
```

```python
highest_consumption = np.where(consumption == np.max(consumption))[0] + 1
lowest_consumption = np.where(consumption == np.min(consumption))[0] + 1
print("highest_consumption: ", Highest_consumption)
print("lowest_consumption: ", Lowest_consumption)
```

```python
monthly_consumption = consumption.reshape(4, 30)
month_1 = monthly_consumption[0]
month_2 = monthly_consumption[1]
month_3 = monthly_consumption[2]
month_4 = monthly_consumption[3]
print("Month 1:", month_1)
print("Month 2:", month_2)
print("Month 3:", month_3)
print("Month 4:", month_4)
```

```python
months = [month_1, month_2, month_3, month_4]
print ("\nTotal consumption analysis:")
month_total = [print(month.sum()) for month in months]
print ("\nAverage consumption analysis:")
month_avg = [print(month.mean()) for month in months]
```

```python
weeks = np.array_split(consumption, 17)
print(weeks)

```

```python
cumulative_consumption = np.cumsum(consumption)
print(cumulative_consumption)
print("\nFirst 10 cumulative consumption values:", cumulative_consumption[:10])
```

```python
lower_bound = mean_consumption - 2 * std_consumption
upper_bound = mean_consumption + 2 * std_consumption
print(lower_bound)
print(upper_bound)
anomaly = np.where((consumption < lower_bound) | (consumption > upper_bound))[0] + 1
print("Anomaly detection (low/high):", anomaly)
print(anomaly)
```

```python
#analysis
Across the whole dataset we got to find out that our mean, median and standard deviation are 408, 414, 115 respectively. 
The data shows that the electricity consumption of each months was not that far apart, but the average shows 
that electricity consumption was slightly higher in the first and second month compared to the others. 
Anomaly test also shows there is no anomaly in the dataset which means all data falls within the expected range.
```
